Every value is a string. Storing numeric values as strings is problematic because you can't really easily apply np or astropy functions to strings, so you need to individually change the relevant string to an int or float to make it usable.

```python
for col in df.columns:
    if col != "Planet" and col != "Ring System?" and col != "Global Magnetic Field":
        df[col] = pd.to_numeric(df[col], errors = "coerce")
```

The three different data types in the df is now str, int and float.

A string is a sequence of character objects in an array, which combine into one entire object. An int is a non decimal numerical value, like 1, 2, 3, 0, -1, -2, etc. A float is a value with the ability to have decimal precision, like 1.0, 2.5, 100.2513424.

All the columns are object dtype. I'm assuming that's because I simply ran through the df column as delineated by the column title, which didn't change the dtype of the column itself, only the dtypes of the individual entries?

```python
# Intakes a dataframe (might be redundant, work with me here), 
# the coefficient in front of the unit, the unit type, and 
# both the old column name and the intended new name.

def attach_units(dframe, coeff, unit, old_col_name, new_col_name):
    new_values = []
    for val in dframe[old_col_name]:
        new_values.append(val * coeff * getattr(u, unit))

    dframe[old_col_name] = new_values
    dframe.rename(columns = {old_col_name: new_col_name}, inplace = True)

    return dframe
```

```python
df = attach_units(df, 1e24, "kg", "Mass (10^24kg)", "Mass (kg)")
```
