```python
import matplotlib.pyplot as plt
from astropy.io import fits
import numpy as np
```

```python
header_list = []
data_list = []

for i in range(1, 6):
    path = "M13_000" + str(i) + ".fits"
    with fits.open(path) as hdul:
        header_list.append(hdul[0].header)
        data_list.append(hdul[0].data)
```

```python
figs, axes = plt.subplots(5, 1, figsize = (4, 14))
for i in range(0, 5):
    axes[i].imshow(data_list[i], cmap = 'viridis', origin = 'lower')
    
plt.show()
```

```python
vmin = []
vmax = []
for i in data_list:
    vmin.append(i.min())
    vmax.append(i.max())

figs, axes = plt.subplots(5, 1, figsize = (4, 14))
for i in range(0, 5):
    axes[i].imshow(data_list[i], cmap='viridis', origin = 'lower', vmin = min(vmin), vmax = max(vmax))

plt.show()
```
