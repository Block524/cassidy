```python
class Rectangle:
    def __init__ (self, Width, Length):
        self.Width = Width
        self.Length = Length

    def area (self):
        return self.Width * self.Length
```

```python
rect1 = Rectangle(10,5)
```

```python
print(rect1.area())
```

```python
class Square(Rectangle):
    def __init__ (self, Side_length):
        super().__init__(Side_length, Side_length)
        self.Area = super().area()
```

```python
square1 = Square(5)
```

```python
print(square1.area())
```

```python
print(square1.Area)
```
