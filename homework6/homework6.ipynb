## Prime Star Systems

```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime_row(inp):
    for elem in inp:
        if elem == 1:
            return False
        elif elem == 2 or elem % 2 == 0:
            return False
        for i in range(2, elem):
            if elem % (i) or elem >= (elem ** 0.5):
                return True
    return False

def return_prime_row(arr_obj):
    list_of_rows = []
    for i in arr_obj:
        if is_prime_row(i):
            list_of_rows.append(i)
    return list_of_rows
```

```python
print(return_prime_row(arr))
```

## Let's Play Checkers!

```python
def create_zeroes_square(inp):
    return np.zeros((inp, inp))

checker = create_zeroes_square(8)
print(checker)
```

```python
def make_checkerboard(checker):
    for i in range(checker.shape[0]):
        for k in range(checker.shape[1]):
            if (i + k) % 2 == 0:
                checker[i, k] = 1
    return checker

print(make_checkerboard(checker))
```

```python
def reverse_checkerboard(checker):
    for i in range(checker.shape[0]):
        for k in range(checker.shape[1]):
            if checker[i,k] == 0.:
                checker[i,k] = 1
            elif checker[i,k] == 1.:
                checker[i,k] = 0
    return checker
```

```python
print(reverse_checkerboard(checker))
```

## The Expanding Universe

```python
def expand(arr, num):
    spaces = ' ' * num
    list_w_space = [spaces.join(word) for word in arr]
    return list_w_space
```

```python
universe = np.array(['galaxy', 'clusters'])

print(expand(universe, 1))
print(expand(universe, 2))
```

## Second-Brightest Star

```python
np.random.seed(123)
stars = np.random.randint(500, 2000, (5,5))
print(stars)
```

```python
def secondBrightest(starList):
    outp_list = np.array([])
    for i in starList:
        outp_list = np.append(outp_list,np.max(i))
    return outp_list
```

```python
print(secondBrightest(stars))
```
