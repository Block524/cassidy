```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
print(data.shape)

# There are 2642 rows

print(data)
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
print(magnitude.min())
print(magnitude.max())

# The maximum magnitude indicates the dimmest of stars, while the minimum magnitude indicates the brightest

print(temperature.min())
print(temperature.max())

# The max temp indicates the hottest stars, assumedly the smallest magnitude, and vice versa for the lowest temp
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize = (8,6))

plt.scatter(log_temperature, log_magnitude, label = "Stars")
plt.scatter(log_temperature[np.argmin(magnitude)], log_magnitude[np.argmin(magnitude)], color = "black", marker = "*", s = 300, label = "Dimmest Star")
plt.scatter(log_temperature[np.argmax(magnitude)], log_magnitude[np.argmax(magnitude)], color = "purple", marker = "*", s = 300, label = "Brightest Star")
plt.scatter(log_temperature[np.argmax(temperature)], log_magnitude[np.argmax(temperature)], color = "green", marker = "*", s = 300, label = "Hottest Star")
plt.scatter(log_temperature[np.argmin(temperature)], log_magnitude[np.argmin(temperature)], color = "orange", marker = "*", s = 300, label = "Coldest Star")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.title("Hertzsprung-Russel Diagram")

plt.legend()


plt.show()
```
