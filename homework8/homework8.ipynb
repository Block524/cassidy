```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
amplitude = [0.5, 1, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
fig, axes = plt.subplots(1, 3, figsize = (25,5))

for amp in amplitude:
    axes[0].plot(x, make_sine_wave(x, amp, 1))
axes[0].set_title("Varying Amplitude")

for freq in frequency:
    axes[1].plot(x, make_sine_wave(x, 1, freq))
axes[1].set_title("Varying frequency")

for amp, freq in zip(amplitude, frequency):
    axes[2].plot(x, make_sine_wave(x, amp, freq))
axes[2].set_title("Varying both")


axes[0].set_xlabel("X")
axes[1].set_xlabel("X")
axes[2].set_xlabel("X")
axes[0].set_ylabel("Y")

plt.show()
```
